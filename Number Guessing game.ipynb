```
# Number guessing game
import random
def game():
    EASY_TURN=10
    HARD_TURN=5
    # function to set difficulty like easy or hard
    def set_difficulty(level):
        if level == "easy":
            return EASY_TURN
        elif level == "hard":
            return HARD_TURN
# check the user guess and actual guess high or low or else
    def check_guess(user_guess,actual_guess,attempts):
            if user_guess>actual_guess:
                print("Too high")
# Reduce the 1 attempt of each incorrect guess
                return attempts -1
            elif user_guess<actual_guess:
                print("Too low")
                return attempts -1
            else:
                print('\n')
                print(f"congratulations! You have guessed the number {actual_guess}")
# Welcome to the game
    print("Welcome to Number guessing game")
    print("I am thinking of a number between 1 to 50. Can you guess it?")
# Take input from user to choose the level
    level=input("Choose difficulty. Type easy or hard::").lower()
    # choose a random number between 1 to 50
    actual_guess=random.randint(1,50)
    print(actual_guess)
# if the user input level is not valid
    attempts=set_difficulty(level)
    if attempts != EASY_TURN and attempts != HARD_TURN:
        print(f"You have chosen the {level} wrong difficulty level. Play again")
# loop the guessing
    user_guess=0
    while user_guess!=actual_guess:
        print(f"You have {attempts} attempts to guess the correct number")
        user_guess=int(input("Make a guess:"))
        attempts=check_guess(user_guess,actual_guess,attempts)
        if attempts==0:
            print(f"You have run out of guesses, You lose!")
            print('\n')
            play_again=input("Would you like to play again? Yes or No").lower()
            if play_again=="yes":
                game()
            else:
                print("\n")
                print("Thank you for playing! Bye😊")
                return
        elif user_guess!=actual_guess:
            print("Guess again")
        elif user_guess==actual_guess:
            one_more_guess=input("Would you like to guess one more number? Yes or No").lower()
            if one_more_guess=="yes":
                print("\n")
                print("Yay! Let's start the game 👏")
                game()
            else:
                print("\n")
                print("Thank you for playing. Bye for now")
game()
```
